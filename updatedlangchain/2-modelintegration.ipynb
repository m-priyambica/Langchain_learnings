### Streaming  And Batch

### Streaming

Most of the  models  can stream their  output while  it is being generated.

By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.

Calling stream() returns an iterator  that  yeilds  output chunks as they are produced .

You can use a loop to process each chunk in real-time.

```python

import os
from langchain_groq import ChatGroq

# Step 1: Initialize the model
model = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.environ["GROQ_API_KEY"],
    temperature=0.7
)

# Step 2: Stream the response
for chunk in model.stream(
    "Write me a 200 words paragraph on artificial intelligence"
):
    print(chunk.content, end="", flush=True)
```

```python
for chunk in dog.stream("write me a 200 words paragraph on artificial intelligence"):
    print(chunk.text)
```

```python
for chunk in dog.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="|",flush=True )
```

## Batch


batching  is a  collection of independent  requests to a model can  significantly  improve performance and reduce costs,as the processing can be done  in parallel.

```python

import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

tap = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.getenv("GROQ_API_KEY"),
    temperature=0.3
)
```

```python
responses = tap.batch([
    "why do parrots have colourful features?",
    "how do airplaines fly",
    "what is quantum computing"
])
for answer in responses:
    print(answer)
```

```python
tap.batch([
    "why do parrots have colourful features?",
    "how do airplaines fly",
    "what is quantum computing"],
    config={
        'max_concurrency': 5, #limit to 5 parrallel calls
    }
    )
```
